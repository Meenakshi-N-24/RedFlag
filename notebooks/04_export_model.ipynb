```python
import re, joblib, sklearn, pandas as pd
from google.colab import files
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

def clean(t):
    t = t.lower()
    t = re.sub(r"https?://\S+|www\.\S+", " urltoken ", t)
    t = re.sub(r"\b\d{10,}\b", " phonetoken ", t)
    t = re.sub(r"[\u20b9$\u00a3]\s?\d[\d,\.]*|\brs\.?\s?\d[\d,\.]*", " moneytoken ", t)
    t = re.sub(r"\d+", " numtoken ", t)
    return re.sub(r"\s+", " ", t).strip()

train_df, val_df = pd.read_csv("train.csv"), pd.read_csv("val.csv")
indian = pd.read_csv("indian_test_set.csv")
indian["y"] = indian["label"]
indian["clean"] = indian["text"].apply(clean)
final = pd.concat([train_df[["clean", "y"]], val_df[["clean", "y"]]] + [indian[["clean", "y"]]] * 10)

vec = TfidfVectorizer(ngram_range=(1, 2), min_df=2, sublinear_tf=True)
clf = LogisticRegression(max_iter=1000, class_weight="balanced", random_state=42)
clf.fit(vec.fit_transform(final["clean"]), final["y"])

joblib.dump(vec, "final_tfidf.joblib")
joblib.dump(clf, "final_lr.joblib")
print("Rows:", len(final), "| scikit-learn version:", sklearn.__version__)
files.download("final_tfidf.joblib")
files.download("final_lr.joblib")
```
